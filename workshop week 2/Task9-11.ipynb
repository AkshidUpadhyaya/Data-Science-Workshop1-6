Task 9: 5x5 random integer matrix (1–100) and max/min

```python
# Generate 5x5 matrix with random integers from 1 to 100
random_ints = np.random.randint(1, 101, size=(5, 5))
print("Task 9 Matrix:\n", random_ints)

print("Maximum value:", random_ints.max())
print("Minimum value:", random_ints.min())
```

Task 10: 4x6 random matrix, row-wise and column-wise sum

```python
# Generate a 4x6 matrix of random decimal numbers
random_matrix = np.random.rand(4, 6)
print("\nTask 10 Matrix:\n", random_matrix)

# Compute sums across axes
row_sums = random_matrix.sum(axis=1)
col_sums = random_matrix.sum(axis=0)

print("Row-wise sum (axis=1):", row_sums)
print("Column-wise sum (axis=0):", col_sums)
```

Task 11: 3x4 random float matrix, mean across axis 0 and axis 1

```python
# Generate a 3x4 matrix of random floats
# (np.random.rand generates floats by default)
float_matrix = np.random.rand(3, 4)
print("\nTask 11 Matrix:\n", float_matrix)

# Compute means across axes
mean_axis_0 = float_matrix.mean(axis=0) # Mean of each column
mean_axis_1 = float_matrix.mean(axis=1) # Mean of each row

print("Mean across axis 0 (columns):", mean_axis_0)
print("Mean across axis 1 (rows):", mean_axis_1)
```
