Task 12: 4x4 matrix to 1D array and shape verification

```python
# Create a 4x4 matrix (using numbers 0 to 15)
matrix_4x4 = np.arange(16).reshape(4, 4)
print("Original shape:", matrix_4x4.shape)

# Convert to 1D
flat_array = matrix_4x4.flatten()
print("New shape after flatten:", flat_array.shape)
```

Task 13: Even numbers (10-100), count, and average

```python
# arange(start, stop, step) - stopping at 101 to include 100
evens = np.arange(10, 101, 2)

count = evens.size
average = evens.mean()

print("\nTask 13 - Total count:", count)
print("Task 13 - Average:", average)
```

Task 14: Variance and Standard Deviation

```python
# Generate 50 random numbers
random_50 = np.random.rand(50)

variance = random_50.var()
std_dev = random_50.std()

print("\nTask 14 - Variance:", variance)
print("Task 14 - Standard Deviation:", std_dev)
```

Task 15: Matrix Transpose and Shape Comparison

```python
# Create a 3x5 matrix
matrix_3x5 = np.random.rand(3, 5)
print("\nTask 15 - Original Shape:", matrix_3x5.shape)

# Transpose it
transposed_matrix = matrix_3x5.T
print("Task 15 - Transposed Shape:", transposed_matrix.shape)
```

Task 16: Array Equality and Greater-Than Count

```python
# Create two random arrays of size 20 (integers 1 to 10)
arr1 = np.random.randint(1, 10, 20)
arr2 = np.random.randint(1, 10, 20)

# Check if exactly equal
are_equal = np.array_equal(arr1, arr2)
print("\nTask 16 - Are arrays exactly equal?:", are_equal)

# Count elements greater in arr1 than arr2
# arr1 > arr2 creates an array of True/False. sum() adds up the Trues.
greater_count = np.sum(arr1 > arr2)
print("Task 16 - Elements greater in first array:", greater_count)
```
