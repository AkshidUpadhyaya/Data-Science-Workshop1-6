```python
import pandas as pd
import matplotlib.pyplot as plt
```

Task: 3 Subplots of Climate Data

```python
# 1. Load the data and convert dates
df = pd.read_csv('TimeSeries Data.csv')
df['date'] = pd.to_datetime(df['date'])

# Create a large figure to hold 3 graphs stacked vertically
plt.figure(figsize=(10, 8))

# --- Subplot 1: Temperature ---
plt.subplot(3, 1, 1) # 3 Rows, 1 Column, 1st Plot
plt.plot(df['date'], df['meantemp'], color='red')
plt.title("Mean Temperature over Time")
plt.ylabel("Temp (C)")

# --- Subplot 2: Humidity ---
plt.subplot(3, 1, 2) # 3 Rows, 1 Column, 2nd Plot
plt.plot(df['date'], df['humidity'], color='blue')
plt.title("Humidity over Time")
plt.ylabel("Humidity (%)")

# --- Subplot 3: Wind Speed ---
plt.subplot(3, 1, 3) # 3 Rows, 1 Column, 3rd Plot
plt.plot(df['date'], df['wind_speed'], color='green')
plt.title("Wind Speed over Time")
plt.xlabel("Date")
plt.ylabel("Wind Speed")

# Adjust spacing so titles don't overlap
plt.tight_layout()
plt.show()
```

Task: Combined Climate Data Plots

```python
plt.figure(figsize=(12, 5))

# Plot 1: Temperature & Humidity ONLY
# (We use a subset of the data, like the first 100 days, so the lines are easier to see!)
subset = df.head(100)

plt.plot(subset['date'], subset['meantemp'], label='Temperature', color='red', linewidth=2)
plt.plot(subset['date'], subset['humidity'], label='Humidity', color='blue', linewidth=2)

plt.title("Temperature and Humidity (First 100 Days)")
plt.xlabel("Date")
plt.ylabel("Values")
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

Task: Temp, Humidity, and Wind Speed Together

```python
plt.figure(figsize=(12, 5))

# Plot all three on the same graph with different colors
plt.plot(subset['date'], subset['meantemp'], label='Temperature', color='red')
plt.plot(subset['date'], subset['humidity'], label='Humidity', color='blue')
plt.plot(subset['date'], subset['wind_speed'], label='Wind Speed', color='green')

plt.title("Temp, Humidity, and Wind Speed (First 100 Days)")
plt.xlabel("Date")
plt.ylabel("Values")
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

Task 2.1: Bar Chart of Avg Temp per Month

```python
# 1. Extract the month into a brand new column
df['month'] = df['date'].dt.month

# 2. Group by the new month column and calculate the average (mean) temperature
avg_temp_month = df.groupby('month')['meantemp'].mean()

# 3. Plot it!
plt.figure(figsize=(8, 4))
plt.bar(avg_temp_month.index, avg_temp_month.values, color='coral')

plt.title("Average Temperature per Month")
plt.xlabel("Month (1 = Jan, 12 = Dec)")
plt.ylabel("Avg Temperature (°C)")
plt.xticks(range(1, 13)) # Forces the x-axis to show all numbers 1-12

plt.show()
```

Task 2.2: Histogram of Temperature Values

```python
plt.figure(figsize=(8, 4))

# We use bins=20 to break the temperatures into 20 even groups
plt.hist(df['meantemp'], bins=20, color='purple', edgecolor='black')

plt.title("Distribution of Temperature Values")
plt.xlabel("Temperature (°C)")
plt.ylabel("Frequency (Number of Days)")

plt.show()
```

Task 2.3: Bar Chart of Avg Humidity per Year

```python
# 1. Extract the year into a brand new column
df['year'] = df['date'].dt.year

# 2. Group by the year and calculate the average humidity
avg_humidity_year = df.groupby('year')['humidity'].mean()

plt.figure(figsize=(8, 4))

# We convert the years to strings (text) so Matplotlib treats them as 
# distinct categories, rather than a continuous math number line!
plt.bar(avg_humidity_year.index.astype(str), avg_humidity_year.values, color='teal')

plt.title("Average Humidity per Year")
plt.xlabel("Year")
plt.ylabel("Avg Humidity (%)")

plt.show()
```
