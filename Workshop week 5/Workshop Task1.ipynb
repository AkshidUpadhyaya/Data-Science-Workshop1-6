```python
import matplotlib.pyplot as plt
```

Task 1.1: Simple Bar Chart

```python
classes = ['A', 'B', 'C', 'D', 'E']
students = [25, 30, 22, 28, 35]

plt.figure(figsize=(6, 4)) # Optional: Make the figure a good size
plt.bar(classes, students, color='skyblue')

plt.title("Number of Students per Class")
plt.xlabel("Classes")
plt.ylabel("Number of Students")
plt.show() # Render the first plot
```

Task 1.2: Subplots (3 Bar Charts in 1 Figure)

```python
# The Data (Re-defining classes just in case)
classes = ['A', 'B', 'C', 'D', 'E']
sec1_students = [25, 30, 22, 28, 35]
sec2_students = [20, 27, 25, 30, 32]
sec3_students = [18, 24, 20, 26, 29]

# Create a wide figure to fit 3 plots side-by-side
plt.figure(figsize=(15, 4)) 

# --- Plot 1 (Section 1) ---
plt.subplot(1, 3, 1) 
plt.bar(classes, sec1_students, color='lightgreen')
plt.title("Section 1 Students")
plt.xlabel("Classes")
plt.ylabel("Students")

# --- Plot 2 (Section 2) ---
plt.subplot(1, 3, 2) 
plt.bar(classes, sec2_students, color='orange')
plt.title("Section 2 Students")
plt.xlabel("Classes")

# --- Plot 3 (Section 3) ---
plt.subplot(1, 3, 3) 
plt.bar(classes, sec3_students, color='lightblue')
plt.title("Section 3 Students")
plt.xlabel("Classes")

plt.tight_layout()
plt.show()
```

Task 2: Scatter Plot (Hours vs Marks)

```python
hours = [2, 3, 4, 5, 6, 7]
marks = [40, 50, 55, 65, 70, 80]

plt.figure(figsize=(6, 4))

# Create the scatter plot
plt.scatter(hours, marks, color='purple', marker='o')

# Add labels and title
plt.title("Correlation: Hours Studied vs. Marks Obtained")
plt.xlabel("Hours Studied")
plt.ylabel("Marks Obtained")

# Add a grid to make it easier to read the exact coordinates
plt.grid(True, linestyle='--', alpha=0.6)

plt.show()
```

Task 3: Pie Chart (Fruits Sold)

```python
fruits = ['Apple', 'Banana', 'Orange', 'Mango']
quantity = [50, 30, 15, 5]

# Define custom colors for each fruit
custom_colors = ['#ff9999', '#ffe599', '#ffcc99', '#ffb3e6']

plt.figure(figsize=(6, 6))

# Create the pie chart with all requested parameters
plt.pie(quantity, 
        labels=fruits, 
        autopct='%1.1f%%', # Shows percentages with 1 decimal place
        colors=custom_colors,
        startangle=90,     # Rotates the start of the pie for a better look
        shadow=True)       # Adds a nice 3D shadow

# Add the required title
plt.title("Proportion of Fruits Sold")

plt.show()
```
