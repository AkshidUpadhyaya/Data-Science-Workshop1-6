```python
# Load the CSV
df = pd.read_csv('Water-potability.csv')
```

```python
# Rename the 'ph' column to 'pH'
df = df.rename(columns={'ph': 'pH'})
```

```python
# Convert all data into floats (decimals)
df = df.astype(float)
```

TASK 3: Basic Information

```python
print("\n--- Summary Statistics ---")
print(df.describe())

print("\n--- Missing Values Count ---")
print(df.isna().sum())
```

```python
import numpy as np
```

TASK 4: Handling Missing Values

```python
# 1. Replace all 0 values with NaN in every column EXCEPT the last one ('Potability')
# Get all column names except the last one
cols_to_replace = df.columns[:-1]
```

```python
# Replace 0 with np.nan only in those specific columns
df[cols_to_replace] = df[cols_to_replace].replace(0, np.nan)
```

```python
# 2. Fill missing values using the column mean
# We calculate the mean of each column and fill the NaNs with it
df = df.fillna(df.mean())

print("Missing values after filling with mean:")
print(df.isna().sum())
```

TASK 5: Column Operations

```python
# 1. Remove the third column (index 2) and first three rows (indices 0, 1, 2)
# Drop 3rd column (axis=1)
df = df.drop(df.columns[2], axis=1)
```

```python
# Drop 1st three rows (axis=0)
df = df.drop([0, 1, 2], axis=0)
```

```python
# 2. Select specific columns and use masking
subset_cols = df[['Sulfate', 'Conductivity', 'Turbidity']]
```

```python
# Example Masking: Let's filter this subset to only show rows where Sulfate is > 300
masked_data = subset_cols[subset_cols['Sulfate'] > 300]
print("\n--- Masked Data (Sulfate > 300) ---")
print(masked_data.head())
```

TASK 6: Remove Data Problems

```python
# 1. Remove duplicate rows
df = df.drop_duplicates()
```

```python
# 2. Remove rows that contain missing values 
# (Note: We already filled them with the mean, but this is good practice to catch any leftovers!)
df = df.dropna()
```

```python
print("\n--- Final Dataset Shape after Cleaning ---")
print(df.shape)
```

TASK 7: Filtering Data

```python
# 1. Show rows where Conductivity > 300
high_conductivity = df[df['Conductivity'] > 300]
print("--- High Conductivity (> 300) ---")
print(f"Found {len(high_conductivity)} rows.")
print(high_conductivity.head(3))
```

```python
# 2. Show rows where pH < 7
acidic_water = df[df['pH'] < 7]
print("\n--- Acidic Water (pH < 7) ---")
print(f"Found {len(acidic_water)} rows.")
print(acidic_water.head(3))
```

TASK 8: Sorting Data

```python
# 1. Sort dataset by pH (ascending)
sorted_by_ph = df.sort_values(by='pH', ascending=True)
```

```python
# 2. Sort dataset by Turbidity (descending)
sorted_by_turbidity = df.sort_values(by='Turbidity', ascending=False)
```

TASK 9: Pivot Table

```python
# Let's see the average pH and Hardness grouped by whether the water is safe (Potability 0 vs 1)
pivot_table = pd.pivot_table(
    data=df, 
    values=['pH', 'Hardness'], 
    index='Potability', 
    aggfunc='mean'
)

print("\n--- Pivot Table (Averages by Potability) ---")
print(pivot_table)
```

TASK 10: Export Data

```python
# Save the cleaned dataset into your working directory
# We use index=False so it doesn't write the row numbers into the file
final_filename = 'Cleaned_Water_Potability.csv'
df.to_csv(final_filename, index=False)

print(f"\nSuccess! Cleaned data saved as '{final_filename}' in {os.getcwd()}")
```
