```python
# Recursive function to calculate factorial
def calculate_factorial(n):
    # Base case: the factorial of 0 or 1 is 1
    if n == 0 or n == 1:
        return 1
    # Recursive case: n * factorial of (n-1)
    else:
        return n * calculate_factorial(n - 1)
```

```python
# Error handling block
try:
    # Ask the user for input and try to convert it to an integer
    user_input = int(input("Enter a positive integer to calculate its factorial: "))
    
    if user_input < 0:
        print("Error: negative numbers cannot be input.")
    else:
        result = calculate_factorial(user_input)
        print("The factorial of ", user_input," is ", result)
        
except ValueError:
    # This block catches errors if the user types a letter or decimal
    print("Invalid input! Please enter a whole number.")
```
