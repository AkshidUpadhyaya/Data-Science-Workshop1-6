```python
import sqlite3
import pandas as pd
```

```python
# 1. Create a physical database named "Employee" (this creates an Employee.db file)
conn = sqlite3.connect('Employee.db')

# A cursor is like a messenger. You give the cursor your SQL commands, 
# it takes them to the database, executes them, and brings the results back.
cursor = conn.cursor()
```

```python
# 2. Create the Employee Table (EmpTable)
# CREATE TABLE defines the schema (the columns and their data types).
# IF NOT EXISTS is a safety feature so the program doesn't crash if you run it twice.
# PRIMARY KEY means 'Empno' must be unique for every single person.
cursor.execute("""
CREATE TABLE IF NOT EXISTS EmpTable (
    Empno INTEGER PRIMARY KEY,
    Ename TEXT,
    Designation TEXT,
    Deptno INTEGER,
    DOJ TEXT,
    BasicSalary INTEGER
)
""")
# commit() saves the changes to the actual physical file.
conn.commit()
```

```python
print("Database and table created successfully.")
```

```python
# We delete all existing rows before inserting new ones to avoid duplicate errors 
# when re-running this notebook cell.
cursor.execute("DELETE FROM EmpTable")
```

```python
# 2. INSERTING DATA
```

```python
# We store our sample data as a list of tuples (each tuple is one row/employee).
employees = [
    (101, 'Ajay', 'Manager', 10, '2001-12-12', 25000),
    (102, 'Arvind', 'Dy. Manager', 40, '2002-02-20', 20000),
    (103, 'Rahul', 'Programmer', 20, '2001-03-22', 18000),
    (104, 'Hari', 'Mktg. Officer', 40, '2001-04-01', 15000),
    (105, 'Sahil', 'Accountant', 10, '2000-09-20', 12000)
]
```

```python
# executemany() runs the INSERT command for every item in our list.
# The question marks (?) act as secure placeholders for the data to prevent errors.
cursor.executemany("INSERT INTO EmpTable VALUES (?, ?, ?, ?, ?, ?)", employees)
conn.commit()
print("Records inserted successfully.\n")
```

```python
# 3. EXECUTING QUERIES & EXPLANATIONS
```

```python
print("1. Display all records:")
# SELECT * means "retrieve every single column" from the table.
cursor.execute("SELECT * FROM EmpTable")
# fetchall() grabs all the rows the cursor found and lets us loop through them.
for row in cursor.fetchall():
    print(row)
```

```python
print("\n2. List names and designations:")
# Instead of *, we specify exact column names to only retrieve that specific data.
cursor.execute("SELECT Ename, Designation FROM EmpTable")
for row in cursor.fetchall():
    print(row)
```

```python
print("\n3. Show employees in department 40:")
# WHERE acts as a filter. It only keeps rows where the Deptno matches 40 exactly.
cursor.execute("SELECT * FROM EmpTable WHERE Deptno = 40")
for row in cursor.fetchall():
    print(row)
```

```python
print("\n4. Total number of employees:")
# COUNT(*) is an aggregate function. It counts the total number of rows.
cursor.execute("SELECT COUNT(*) FROM EmpTable")
# fetchone() grabs just the first row of results. 
# We use [0] to extract the actual number from the tuple it returns (e.g., extracting 5 from (5,)).
print(cursor.fetchone()[0])
```

```python
print("\n5. Names of employees with salary > 18000:")
# The > operator filters for strictly greater values.
cursor.execute("SELECT Ename FROM EmpTable WHERE BasicSalary > 18000")
for row in cursor.fetchall():
    print(row[0])
```

```python
print("\n6. Employee details who joined before 2002:")
# You can use standard math operators (<, >) on dates if they are formatted YYYY-MM-DD.
cursor.execute("SELECT * FROM EmpTable WHERE DOJ < '2002-01-01'")
for row in cursor.fetchall():
    print(row)
```

```python
print("\n7. Highest salary among employees:")
# MAX() scans the entire BasicSalary column and returns only the single highest number.
cursor.execute("SELECT MAX(BasicSalary) FROM EmpTable")
print(cursor.fetchone()[0])
```

```python
print("\n8. Names of employees whose salary is between 15000 and 25000:")
# BETWEEN creates an inclusive range. It is identical to writing: BasicSalary >= 15000 AND BasicSalary <= 25000.
cursor.execute("SELECT Ename FROM EmpTable WHERE BasicSalary BETWEEN 15000 AND 25000")
for row in cursor.fetchall():
    print(row[0])
```

```python
print("\n9. Employees who are 'Programmer' or 'Accountant':")
# IN () is a shortcut. It checks if the designation matches any item inside the parentheses.
cursor.execute("SELECT * FROM EmpTable WHERE Designation IN ('Programmer', 'Accountant')")
for row in cursor.fetchall():
    print(row)
```

```python
print("\n10. Total salary paid to all employees:")
# SUM() adds together every single number in the BasicSalary column.
cursor.execute("SELECT SUM(BasicSalary) FROM EmpTable")
print(cursor.fetchone()[0])
```

```python
print("\n11. Employees who joined in the year 2001:")
# We use BETWEEN again to capture any date from January 1st to December 31st of that year.
cursor.execute("SELECT * FROM EmpTable WHERE DOJ BETWEEN '2001-01-01' AND '2001-12-31'")
for row in cursor.fetchall():
    print(row)
```

```python
print("\n12. Details sorted by salary in descending order:")
# ORDER BY sorts the final results. DESC forces it to go from highest to lowest. 
# (If we wanted lowest to highest, we would use ASC).
cursor.execute("SELECT * FROM EmpTable ORDER BY BasicSalary DESC")
for row in cursor.fetchall():
    print(row)
```

```python
print("\n13. Average salary in department 40:")
# AVG() calculates the mathematical mean of the salary column, but only for the rows that pass the WHERE filter.
cursor.execute("SELECT AVG(BasicSalary) FROM EmpTable WHERE Deptno = 40")
print(cursor.fetchone()[0])
```

```python
print("\n14. Employee with the lowest salary:")
# This uses a "Subquery" (a query inside a query).
# The database first runs the inner query: (SELECT MIN(BasicSalary) FROM EmpTable) which evaluates to 12000.
# Then it runs the outer query: SELECT * FROM EmpTable WHERE BasicSalary = 12000.
cursor.execute("SELECT * FROM EmpTable WHERE BasicSalary = (SELECT MIN(BasicSalary) FROM EmpTable)")
for row in cursor.fetchall():
    print(row)
```

```python
print("\n15. Joined after 01-01-2001 and salary > 15000:")
# The AND keyword allows you to combine multiple conditions. Both rules must be true for a row to be retrieved.
cursor.execute("SELECT * FROM EmpTable WHERE DOJ > '2001-01-01' AND BasicSalary > 15000")
for row in cursor.fetchall():
    print(row)
```

```python
# Safely close the connection when the program is completely finished to avoid locking the file.
conn.close()
```
